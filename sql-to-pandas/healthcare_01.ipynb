# ER Mean & Median Wait Times: SQL to Pandas
Calculating average and median wait times from emergency room records. Validating SQL findings with Python for accuracy. 

```python
import pandas as pd
```

## Load ER Visits Data

```python
er_visits = pd.read_csv("../data/raw/er_visits.csv") 
```

## Preview of ER Visits Table

```python
er_visits.head()
```

## SQL to Pandas Validations: Mean & Median ER Wait Times

```python
round(er_visits["wait_time_minutes"].mean(),2)
```

```python
er_visits["wait_time_minutes"].median()
```

## Breakdown of the Validation Results 
The average ER wait time is 74.31 mintues, and the median wait time is 67 minutes. Both results validate the same findings from SQL.
