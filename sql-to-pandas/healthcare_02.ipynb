# ER Wait Times by Severity Level 
Calculating average ER wait times by severity level. Validating SQL findings with Pandas for accuracy. 

```python
import pandas as pd
```

## Load ER Visits Data

```python
er_visits = pd.read_csv("../data/raw/er_visits.csv") 
```

## Preview of ER Visits DataFrame

```python
er_visits.head()
```

## SQL to Pandas Validation: Average ER Wait Times by Severity Level 

```python
er_visits.groupby("severity_level")["wait_time_minutes"].mean().sort_values(ascending=False)
```

## Breakdown of Validation Results
The higher the severity level of the appoinment, the shorter the overall wait time. Low Severity pts are seen with avg wait time of 109 min
Med Severity are pts seen with avg wait time of 73 min, High Severity pts are seen with avg wait time of 35 min, and Critical pts seen with avg wait time of 7 min.
